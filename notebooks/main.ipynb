```python
from inference import predict_muac_from_path, IMAGES_DIR
from calculator import classify_nutrition_status
import os
```

```python
def assess_child_from_image(
    age_months: float,
    weight_kg: float,
    height_cm: float,
    image_filename: str
):
    img_path = os.path.join(IMAGES_DIR, image_filename)

    muac_pred = predict_muac_from_path(img_path)

    status, explanation, advice = classify_nutrition_status(
        age_months=age_months,
        weight_kg=weight_kg,
        height_cm=height_cm,
        muac_cm=muac_pred
    )

    print(f"Image         : {image_filename}")
    print(f"Predicted MUAC: {muac_pred:.2f} cm")
    print(f"Status        : {status}")
    print()
    print("Explanation:")
    print(explanation)
    print()
    print("Advice:")
    print(advice)

    return {
        "image": image_filename,
        "muac_pred_cm": muac_pred,
        "status": status,
        "explanation": explanation,
        "advice": advice,
    }
```

```python
result = assess_child_from_image(
    age_months=24,
    weight_kg=10.5,
    height_cm=82.0,
    image_filename="Normal_3.jpg"  # ganti sesuai nama file di dataset/images
)
```
